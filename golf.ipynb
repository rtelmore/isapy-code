## If you don't know anything about golf

Some of you may not be as familiar with golf as the other sports mentioned in this text. Golf differs from these other sports in many ways. First, golf is an individual sport, not a team sport like the other sports considered in individual chapters in this text. Second, the participants compete against each other only indirectly: Each participant plays the golf course and the actions of any one player do not change the ability of any other player to play the shot he or she wants to play. Third, at the highest levels of the game and at courses in cities every weekend, the players generally compete in a tournament format. That is, they do not compete one-on-one with a win or a loss, but rather scores are aggregated across the entire tournament to determine the winner and subsequent positions. In any given week on the PGA Tour, you could say that there is one tournament winner and 155 "losers", but rather players determine their relative performance by position finished, such as a runner up, top-10 finish, or perhaps a top-25 finish.

Yet in other ways, golf is similar to the sports we have studied elsewhere in this book. For example, each stroke in golf is a discrete event, much like a pitch in baseball or a play in football. It is played exclusively outdoors, meaning it is subject to the weather conditions that you see in sports like baseball and soccer. The strategy that a player selects for his/her own shot, combined with its execution, will determine the score a player achieves on any given hole. When we examine the Strokes Gained metrics later in this chapter, we will see how these decisions, and their results, compound to create a score on each hole, much like the strategy used in American football when choosing conservative or aggressive strategies on offense and defense.

In case you are not familiar with some of the terminology used here, we refer you to the Wikipedia page for the Glossary of Golf, available at [https://en.wikipedia.org/wiki/Glossary_of_golf](https://en.wikipedia.org/wiki/Glossary_of_golf).

## The Evolution of Golf Analytics

Golf did not have the long history of analytics the way that a sport like baseball did. Sportswriters do not create box scores for golf the way they did for baseball or football. This may be due to the relatively simple nature of the game (low score wins) and the small numbers being easy to count. Moreover, through the first half of the 20th century the most popular form of the game at the professional level (and still the most popular form today at weekend games between friends at golf clubs) was match play, where one player competes against one other player, and the winner of the match is the player who has the lowest score on the largest number of holes, rather than the lowest total score over all 18 holes.

Due to the above, early golf metrics were very simple. Usually, the best professional player of the year was judged by the amount of prize money won that year, or perhaps the number and importance of tournaments won, and sometimes even by just a vote of his/her peers. But any statistics beyond those were generally not kept.

In 1980, the PGA Tour began to publish statistics about the players in their events. While these are basic by today's standards, at the time it was considered a revolutionary way for players to be compared and evaluated in several different ways instead of just score on a hole. These statistics included driving distance, driving accuracy, greens in regulation percentage, scrambling percentage, sand save percentage, putts per round, and par breakers. So many of these statistics have their inherent flaws when viewed in isolation. For example, one player might have very few putts per round, and so it looks as though this person is a superior putter. One explanation could be, however, that this player was not very good at hitting the green from the fairway, and as such chipped a lot. It is much easier to get the ball quite close to the hole when chipping from 20 yards than when hitting from 200 yards, and as such the closer the ball is to the hole, the more easily explained the small number of putts[^golf-1].

[^golf-1]: Famously in the final tournament round of 1984, PGA Tour player and later television announcer Gary McCord, who had over 400 PGA Tour starts but 0 wins, intentionally attempted to miss all 18 greens in regulation with his approach shots in order to increase his chances of leading the average putts per round statistic for the year.  He was successful both in missing every green that day and in subsequently winning the average putts per round (28.57) statistical category for the year over Hubert Green (28.69).

## Advanced Golf Metrics

In 2001, the PGA Tour began the Shotlink program to more accurately track each stroke that players make in PGA Tour tournaments. Operated largely by volunteers, three people would shoot a laser at the ball after it has come to rest. This data would then be fed into computers which would triangulate the ball's position relative to the hole and the location of the previous shot to determine how far that shot traveled, its condition (fairway, rough, sand, green, etc.), and how far the ball now remains from the hole. This enabled far more accurate tracking of distances than previously used for simply measuring driving distance (which was only done on two holes per tournament) and enabled the announcers to have more accurate data as well. The next generation Shotlink+ is adding camera data in and around greens to gather even more granular data.

```python
#| fig-cap: "Shotlink camera"
#| label: fig-camera
#| echo: false

display(Image(filename="fig/shotlink.jpeg"))
```

With the advent of Shotlink came a whole new era in golf statistics, under the umbrella of Strokes Gained. Strokes Gained calculations allow a comparison for how many strokes someone is expected to take to get the ball in the hole from a certain location and lie, such as from 400 yards away off of the tee, 170 yards away in the fairway, or 110 yards out in the rough, based on a baseline established from data collected on all PGA Tour players. For an in-depth explanation of the formulas used, see https://golfity.com/strokes-gained/.

Mark Broadie at Columbia University is credited with first creating the strokes gained concept. He used this as a way to get rid of the complicating factors mentioned in the previous section on one category of data unduly influencing another category. Professor Broadie has since become well known in the golf analytics community, authoring the book *Every Shot Counts*, writing a monthly column for Golf magazine and working with the United States Golf Association on their handicap research team.

### Calculating Strokes Gained {.unnumbered}

To explain strokes gained in statisticians' terms, we will define a function $J(d,c)$. This function represents the average number of strokes it takes a PGA Tour golfer to finish a hole, considering the distance from the hole $d$, and the condition of the stroke about to be played $c$. To calculate the strokes gained on a given shot $i$, we want to take the difference between the average number of shots it takes a player to complete the hole before shot $i$ and after shot $i$. We then subtract one from that difference because the player just completed a shot. Represented mathematically, the equation is
$$
g_i = J(d_i,c_i) - J(d_{i+1}, c_{i+1}) - 1.
$$ As an example, let us borrow from the explanations on [PGATour.com](PGATour.com) and consider tour player Rickie Fowler on the 18th (final) hole in the 4th (final) round of The Players Championship in 2015. The hole is a 446 yard par four, and the PGA Tour average for a 446 yard hole is 4.1 strokes.

**Shot 1**: Fowler hit his tee shot 330 yards in the fairway, leaving himself 116 yards to the cup. From 116 yards in the fairway, the average tour player needs 2.825 strokes to complete the hole. So the calculation of his strokes gained on the tee shot is $$
g=4.1 -2.825-1= +0.275
$$ **Shot 2**: Fowler then hit his shot from 116 yards in the fairway onto the green, to a distance of 16 feet 11 inches from the cup. The average tour player needs 1.826 shots from a distance of 16' 11" on the green. Therefore, his strokes gained on the second shot is $$
g=2.825 -1.826-1= -0.001.
$$ **Shot 3**: Fowler then proceeded to make the putt (and win the tournament later in a playoff). He has no more strokes to take once the ball is in the cup. Thus we calculate his strokes gained putting as $$
g=1.826 -0-1= + 0.826.
$$ We can calculate the strokes gained or lost on every hole, or for the round, or for the year. Often, Strokes Gained when evaluating a player are grouped into four categories

1.  Strokes Gained Off the Tee (SG OTT): Tee shots on par 4 and par 5 holes.
2.  Strokes Gained Approach (SG APP): All shots not from the tee on par 4 and par 5 holes that are not included in Strokes Gained Around The Green. Also includes tee shots on par 3 holes.
3.  Strokes Gained Around The Green (SG ARG): Any shot that happens within 30 yards of the edge of the green but is not on the green.
4.  Strokes Gained Putting (SG PUTT): Shots from the putting surface.

You will often see 1 - 3 grouped as Strokes Gained Tee-to-Green.

Obviously the number of permutations with these statistics are endless. For example, a player wanting to study how he is performing on shots with full-swing wedges relative to the field might look at Strokes Gained 100-150 yards.

We will use Python to analyze data and display it in a radar chart to show the relative importance of the four types of shots gained data relative to player success in PGA Tour events for the 2022 season right up to the split with LIV Golf as a competing tour.

To do this, our script will use `pandas` for data retrieval and manipulation, and `matplotlib` for graphing the data in a radar plot. We will use two data sets from the `isapy_data` module: `ow_golf_rankings` (Official World Golf Rankings) and `pga_tournaments` (from Datagolf). First, we will merge the two data sets using the `merge()` method.

```python
pga_tournaments = isapy_data.load_data("pga_tournaments")
ow_golf_rankings = isapy_data.load_data("ow_golf_rankings")

df = pga_tournaments.merge(
    ow_golf_rankings, on="Player_initial_last", how="left"
)
```

Exploring the data, we see that for each tournament we have player data, such as name, whether he made the 36-hole cut or not, total number of strokes taken in the tournament, fantasy points earned in various competitions by Draftkings, FanDuel, and Showdown, and finishing position in the tournament, all merged with the OWGR ranking from above. We clean this data a bit by assigning anyone who was not in the top 300 in the OWGR a ranking of 301. Moreover, we see that if a player missed the cut, their finishing position is listed as NaN, and we instead assign each NaN a value of 999. We also remove rows where a player's SG data is for whatever reason missing.

We are mainly interested in the SG data and OWGR data, so we are going to select only those columns into a new DataFrame, and then create another DataFrame which has the averages of each of the 4 main SG categories for each player.

```python
df["WGR_June_2022"] = df["WGR_June_2022"].fillna(301)
df["pos"] = df["pos"].fillna(999)

df_sub = df[["Player_initial_last", "sg_putt", "sg_arg", "sg_app",
             "sg_ott", "sg_t2g", "sg_total", "WGR_June_2022"]].dropna()

df_summarized = (
    df_sub
    .groupby("Player_initial_last")
    .agg(
        mean_sg_ott=("sg_ott", "mean"),
        mean_sg_app=("sg_app", "mean"),
        mean_sg_arg=("sg_arg", "mean"),
        mean_sg_putt=("sg_putt", "mean"),
        mean_WGR=("WGR_June_2022", "mean")
    )
    .reset_index()
)
df_summarized["top_50"] = df_summarized["mean_WGR"] <= 50
```

Our objective in this exercise was to see how the SG stats compared between players who were currently the best in the world and those who were not quite there yet. So we create a flag in the dataset to delineate players as being either in the top 50 or outside the top 50.

Now we summarize the data to get the mean SG scores for the Top 50 players in the OWGR and those who played in the PGA Tour tournaments that were not in the Top 50.

```python
df_summarized_top_50 = (
    df_summarized
    .groupby("top_50")
    .agg(
        SGOTT=("mean_sg_ott", "mean"),
        SGAPP=("mean_sg_app", "mean"),
        SGARG=("mean_sg_arg", "mean"),
        SGPUTT=("mean_sg_putt", "mean")
    )
    .reset_index()
)
df_summarized_top_50
```

Next, we need to prepare the data for the radar chart. In Python, we use `matplotlib`'s polar projection to create radar charts. We need to define the categories (the four SG metrics), compute the angles for each axis, and then plot the values as filled polygons. We close the polygon by appending the first value to the end of each list.

```python
#| label: fig-golf-radar
#| fig-cap: "Radar plot showing four average strokes gained metrics for golfers in the top 50 versus those not in the top 50 of the OWGR."

categories = ["SG OTT", "SG APP", "SG ARG", "SG PUTT"]
N = len(categories)

not_top_50 = df_summarized_top_50[df_summarized_top_50["top_50"] == False]
top_50 = df_summarized_top_50[df_summarized_top_50["top_50"] == True]

sg_cols = ["SGOTT", "SGAPP", "SGARG", "SGPUTT"]
values_not_top50 = not_top_50[sg_cols].values.flatten().tolist()
values_top50 = top_50[sg_cols].values.flatten().tolist()

values_not_top50 += values_not_top50[:1]
values_top50 += values_top50[:1]

angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
ax.set_title("Strokes Gained Metrics for Professional Golfers",
             size=14, pad=20)

ax.fill(angles, values_not_top50, alpha=0.25, color="red",
        label="Not Top 50 Players")
ax.plot(angles, values_not_top50, color="red", linewidth=2)

ax.fill(angles, values_top50, alpha=0.25, color="green",
        label="Top 50 Players")
ax.plot(angles, values_top50, color="green", linewidth=2)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)
ax.legend(loc="upper right", bbox_to_anchor=(1.3, 1.1))

plt.tight_layout()
plt.show()
```

We can now see visually the gaps between the players in the two groups. From looking at the radar chart, it seems that the biggest difference between the groups is in SGAPP, and the smallest is in SGARG. So a player who is ranked #200 and wants to move into the top 50 might decide to spend the majority of his practice time working on his shots from 100-200 yards and less time on chipping and putting.

## Cluster Analysis

Much like we did in the soccer chapter, we can use k-means clustering for unsupervised learning. As a reminder, in k-means clustering, we specify a number of clusters (say $k$ clusters) that potentially underlie our data. We then calculate $k$ centroids (usually group means or medians) and minimize the total distances between the data and the centroids. Given that we have four different dimensions of shots gained data, we can use them as dimensions and see if we can identify which of the shots gained metrics are most important in identifying OWGR ranking of our players.

We will also use `StandardScaler` from `scikit-learn` to standardize the features before clustering, ensuring that all four SG metrics are on the same scale for distance calculations. At this point, though, we will do a few things differently. For example, we may want to group players into the top 50 in the OWGR (usually this will get a player invitations to The Masters and other special events, top 100 in the OWGR (usually this will get a player an invitation to the PGA Championship), and everyone else.

```python
df_clusters = df_summarized[["Player_initial_last", "mean_sg_ott", "mean_sg_app",
                              "mean_sg_arg", "mean_sg_putt"]].dropna()

df_summarized["three_tiers"] = np.where(
    df_summarized["mean_WGR"] <= 50, "top_50",
    np.where(df_summarized["mean_WGR"] <= 100, "top_100", "outside_100")
)
```

```python
features = df_clusters[["mean_sg_ott", "mean_sg_app", "mean_sg_arg", "mean_sg_putt"]]

scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

km = KMeans(n_clusters=3, random_state=90283, n_init=20)
km.fit(features_scaled)

centers_original = scaler.inverse_transform(km.cluster_centers_)
centers_df = pd.DataFrame(
    centers_original,
    columns=["mean_sg_ott", "mean_sg_app", "mean_sg_arg", "mean_sg_putt"],
    index=["Cluster 1", "Cluster 2", "Cluster 3"]
)
centers_df
```

```python
df_cl_results = df_clusters.copy()
df_cl_results["clusters"] = km.labels_ + 1
df_cl_results = df_cl_results.merge(df_summarized, on="Player_initial_last", how="left",
                                     suffixes=("", "_y"))

pd.crosstab(df_cl_results["clusters"], df_cl_results["three_tiers"])
```

One of the first things we may notice is that we only have 48 of the top 50 players and only 37 of the second 50 players. We must have eliminated many of these players when we cleaned the data to remove some of the rows with NA. It is also probable that many of these players did not play in any of the PGA Tour tournaments during this time frame. This could be due to injury or that those players primarily played on other tours like the DP World (European) Tour or Japanese Tour.

This output gives us the three clusters which we requested. From the cluster centers, we can examine how each cluster is characterized by the four strokes gained metrics. We can look at which clusters have positive or negative means in each category to understand the relative strengths and weaknesses of the players in each group. On average, the cluster with the majority of our top players tends to have positive strokes gained in all categories, excelling particularly on approach shots.

## Research Block: Does Par on a Hole Matter?

On a standard golf course, every hole has a par rating.  Par for a given hole is almost always rated as a 3, 4, or 5, which is the number of shots a skilled golfer should expect to take to reach the green, plus two for the number of putts a skilled golfer would expect to take on that green.  Even on minigolf courses in beach resorts one can find each hole rated by the number of putts expected to complete the hole. But what does that number really mean in the grand scheme of things, and does it affect a golfer's psyche?

In competition, par on a hole or for a course does not really matter.  It merely creates a reference point that allows someone to refer to their status at a given point in the tournament not by the number of strokes taken, but in relation to par.  For example, at the 2026 AT&T Pebble Beach Pro-Am, Collin Morikawa was the winner, needing 266 strokes to complete 72 holes.  The next closest competitor needed 267 strokes.  Normally reporters and players alike do not refer to their score by the total number of strokes but by the score in relation to par.  It would be reported that Morikawa was -22 (or 22 under par) and won by one.  For the purposes of the competition, it would not matter if we decided that 266 strokes was -22, -32, or +2, he would still be the winner by completing the 72-hole tournament in the lowest number of strokes.

Research by the textbook authors, published in the paper "Loss Aversion in Professional Golf" in *Journal of Sports Economics*, available at [https://journals.sagepub.com/doi/abs/10.1177/1527002520967403](https://journals.sagepub.com/doi/abs/10.1177/1527002520967403), shows that professional golfers do indeed suffer from the behavioral economic bias of loss aversion.  Professional golfers anchor on whether a hole is rated as a par-5 or a par-4 and exhibit behavior that causes them to try harder to ensure that they score a par as opposed to trying for one stroke below par (known in golf parlance as a birdie).  A natural experiment where the USGA played the US Open on the same golf courses (Pebble Beach in California as well as Oakmont in Pennsylvania) but in one year called the same hole a par-5 as opposed to in the following year calling it a par-4 resulted in players playing the hole approximately .25 strokes better as a par-4, all other things accounted for.  Quotes from several players even indicated that they knew in the grand scheme of things that par didn't matter, and yet it still affected their attitudes towards playing the hole, particularly after hitting a bad shot.

## Lab: Simulating the Cut

Currently, the tournaments on the PGA Tour have a curious financial structure for the participants.
Each player in the tournament covers his own travel, coaching and caddie expenses, and pays a nominal entry fee.
Then, normally after the second day of the four-day tournament, the field of 120-160 players is "cut" to the top 65 players plus any ties. This results in about half of the field coming to play each week earning nothing, and in fact losing money due to ancillary expenses[^golf-2]. The players who survived the cut then play two more days, and the prize money distributed according to their place in the tournament.  It is a top-heavy structure, where normally 18\% of the prize pool goes to the winner, 10.9\% to second place, 6.9\% to third place, and in continuing smaller increments down to 0.215\% for the person finishing $65^{th}$ and beyond.  In a tournament with a \$10 million prize pool, a player would have to finish $23^{rd}$ or better to earn a 1\% payout of \$100,000.

[^golf-2]: The authors would not enjoy teaching against 154 other professors for half the semester to then see who actually gets paid and who gets sent home broke.

In addition to the short-term financial incentives for finishing high in a tournament, there are also many incentives for winning.  A tournament winner not only receives 18\% of the prize pool, but also other incentives such as a 2-year automatic exemption from qualifying for the PGA Tour and automatic entry into other prestigious tournaments for the following year like The Masters.

So when a player is playing on the PGA Tour, he makes various decisions each time he has a shot.  An aggressive strategy to aim directly at a hole that is located near a pond might bring the player a reward of a low score if he executes it perfectly but also brings in danger and penalty strokes if his ball were to land slightly offline in or near the water.

We can use analytics of course to help us make better decisions, but it depends on how one defines "better". One player may choose a strategy with the best chance of making cuts and ensuring that he has a job for the following year, but another player may say he is ignoring the money and playing to win.  This lab helps us to start modeling some of these strategy decisions.

Suppose we have two players, Player X and Player Y. They are professional golfers and both are competing in the National Open.  We want to model their potential outcomes in the tournament.

Player X has probabilities of recording a one, two, three, four, five, and six on any par three hole as 0.0001, 0.25, 0.6, 0.1, 0.04, and 0.0099, respectively. We write this as $\pi_3^X = (0.0001,0.25,0.6,0.1,0.04,0.0099).$ His probability of recording a two through seven on par 4s and par 5s can be written as $\pi_4^X = (.0005,.2,.6,.1,.09,0.0095)$ and $\pi_5^X = (0,0.05,0.35,0.5,0.0901,0.0099),$
respectively. Player Y is a little more volatile and has a slightly higher probability of making a birdie or better, or a bogey or worse, than Player X. His probability vectors can be written as $\pi_3^Y = (0.0001, 0.30, 0.5, 0.14, 0.05, 0.0099)$, $\pi_4^Y = (0.0005, .25, .5, .15, .09, 0.0095)$, and $\pi_5^Y = (0, 0.05, 0.375, 0.45, 0.1151, 0.0099)$ on Par 3s, Par 4s, and Par 5s, respectively.

1. Given that the course used in the National Open tournament has four par 3 holes, ten par 4 holes, and four par 5 holes, which player do you think is more likely to make the cut if we predict that that players need to shoot 146 or lower after two rounds to play the weekend?
2. Which player is more likely to have a better overall (four rounds) score?

For both of the above questions, assume that holes are independent. That is, a player's performance on one hole is independent of his performance on any other hole, including the same hole the next day. Each hole's score can be sampled from a multinomial distribution given the probability vectors defined above.

To do this lab, we will use simulation.  You may recall that we used simulation in the football chapter when we simulated drives based on actual plays from an NFL player database.  We will do this again here, and this time we will draw 10,000 Monte Carlo samples.

First we will define the probability matrices for each player. Each row in the matrix corresponds to the probabilities for par-3, par-4, and par-5 holes. The possible scores on par-3 holes are 1 through 6, while on par-4 and par-5 holes they are 2 through 7.

The for loop that follows runs 10,000, or `B` times, one for each tournament outcome we are simulating. We use `np.random.multinomial()` to sample scores from a multinomial distribution. We sample from the multinomial distribution eight times for the par 3 and par 5 holes, and 20 times for the par 4s. Recall that in the two rounds before the cut in the National Open, each player will play 8 par 3s, 20 par 4s, and 8 par 5s for their 36 holes. We will then sum the scores on those holes, and repeat for each of the 10,000 times. The two players' scores will be recorded in the `x_scores` and `y_scores` arrays.

We then calculate the mean probability of making the cut by counting the number of scores in each array that are less than or equal to our projected cut number (146) and dividing by the number of times the simulation was run. Furthermore, we create a 95% confidence interval for the mean by calculating the standard deviation and adding and subtracting 1.96 standard deviations from our mean.

We can then plot the two distributions on top of each other, and see visually that as expected Player Y has a wider distribution of scores, whereas the more consistent Player X has a narrower distribution.

Before running the simulation, we set a **random seed** with `np.random.seed()`. Because simulation draws random numbers, results will differ slightly every time the code runs. Setting a seed fixes the starting point of the random number generator so that anyone running this code gets identical results — an important practice for reproducibility in research.

```python
np.random.seed(19831)
B = 10000

x_probs = np.array([
    [0.0001, 0.25, 0.6, 0.1, 0.04, 0.0099],
    [0.0005, 0.2, 0.6, 0.1, 0.09, 0.0095],
    [0, 0.05, 0.35, 0.5, 0.0901, 0.0099]
])
y_probs = np.array([
    [0.0001, 0.30, 0.5, 0.14, 0.05, 0.0099],
    [0.0005, 0.25, 0.5, 0.15, 0.09, 0.0095],
    [0, 0.05, 0.375, 0.45, 0.1151, 0.0099]
])

par3_scores = np.arange(1, 7)
par45_scores = np.arange(2, 8)

x_scores = np.zeros(B)
y_scores = np.zeros(B)

for b in range(B):
    threes_x = np.sum(np.random.multinomial(1, x_probs[0], size=8) @ par3_scores)
    fours_x = np.sum(np.random.multinomial(1, x_probs[1], size=20) @ par45_scores)
    fives_x = np.sum(np.random.multinomial(1, x_probs[2], size=8) @ par45_scores)

    threes_y = np.sum(np.random.multinomial(1, y_probs[0], size=8) @ par3_scores)
    fours_y = np.sum(np.random.multinomial(1, y_probs[1], size=20) @ par45_scores)
    fives_y = np.sum(np.random.multinomial(1, y_probs[2], size=8) @ par45_scores)

    x_scores[b] = threes_x + fours_x + fives_x
    y_scores[b] = threes_y + fours_y + fives_y
```

Estimated probability of making the cut for each player can be found as follows.

```python
px = np.sum(x_scores <= 146) / B
print(f"Player X: {px:.4f}")

py = np.sum(y_scores <= 146) / B
print(f"Player Y: {py:.4f}")
```

Are these differences meaningful in a statistical sense? You can easily calculate a 95\% confidence interval using the rationale mentioned above.

```python
ci_x = (px - 1.96 * np.sqrt(px * (1 - px) / B),
        px + 1.96 * np.sqrt(px * (1 - px) / B))
print(f"Player X 95% CI: ({ci_x[0]:.4f}, {ci_x[1]:.4f})")

ci_y = (py - 1.96 * np.sqrt(py * (1 - py) / B),
        py + 1.96 * np.sqrt(py * (1 - py) / B))
print(f"Player Y 95% CI: ({ci_y[0]:.4f}, {ci_y[1]:.4f})")
```

Because these two confidence intervals overlap, they are not statistically different. Let's plot the distributions of the simulations:

```python
df_sim = pd.DataFrame({
    "name": np.repeat(["Player X", "Player Y"], B),
    "scores": np.concatenate([x_scores, y_scores])
})

p = sns.kdeplot(data=df_sim, x="scores", hue="name", fill=True,
                alpha=0.5, palette="Set1")
p.set_xlabel("Scores")
p.set_ylabel("Density")
plt.show()
```

Suppose a total score of seven under par (-7) will give you the lead after two rounds. Which player is more likely to be in the lead? To calculate this, let's count the number of scores in the simulation that were -7 or better, and then divide that sum by the total number of trials in the simulation,

```python
print(f"Player X P(score <= 137): {np.sum(x_scores <= 137) / B:.4f}")
print(f"Player Y P(score <= 137): {np.sum(y_scores <= 137) / B:.4f}")
```

From the results of our simulation, we believe it is more likely that player Y is in the lead after two rounds than player X.

## Case Study: Ranking Golfers from Different Tours Against Each Other

Every day around the world, amateur golfers of different skills compete against each other based on the World Handicap System.  This system requires players to enter their scores into a system every time they play, along with the course they played and from what set of tees they played.  The formula, simplified a bit for the purposes of this exercise, for calculating a player's handicap is as follows.

1. Identify the 20 most recent 18 hole scores that a player has posted
2. Next we need a way to identify how the relative difficulty level of the courses where the rounds were played.  For example, it is probably easier to shoot an 80 at your local town municipal golf course than it is at Augusta National Golf Club.  To do this, the USGA has determined two different ways of assessing a course's difficulty: course rating and slope.  Course rating is the score that a scratch golfer (i.e., a golfer with a handicap of 0) should shoot on average at that course.  For example, the course rating from the Black (back) tees at the University of Denver Golf Club is 72.9, which means that a scratch golfer should shoot on average about 73.  The course rating is heavily influenced by the length.  Each course also has a slope rating, which is to help identify how much more difficult a course is for less skilled golfers than scratch golfers.  Less skilled golfers will be more affected by deep sand bunkers, water hazards, thick grass, narrow fairways, undulating greens, and other similar difficulties than a more skilled player, In an effort to account for these differences, courses also have a slope rating.  The USGA decided that the slope rating for a golf course of average difficulty is 113, and that courses could be as easy as 55 or as difficult as 155, though one will rarely find the extremes.  TPC Sawgrass is a 155 from their tournament tees.  The same University of Denver Golf Club from the Black tees mentioned above has a slope rating of 129.  As a result you will generally find these on the scorecard for the course (and in the USGA database), often with the nomenclature "rating/slope".  For University of Denver Golf Club, on the scorecard next to the information for the black tees you will see "72.9/129".

Now that we have the course rating and slope for the courses played for each of the scores, we can calculate the handicap differential for the round.  The formula for the handicap differential is the difference between the player's score for the day and the course rating, multiplied by 113 and divided by the slope rating for the course.  For example, a player who shot 95 on the aforementioned University of Denver Golf Course from the Black tees, a course rated at 72.9 and slope rating of 129 would have a handicap differential for that round of $(95-72.9) * (113/129) = 19.36$, rounded to 19.4.

3. Identify the best 8 of those last 20 handicap differentials, calculate the mean, and take 96\% of that mean.

So once again a bit simplified, Player A with a handicap of 5 would be expected to beat Player B with a 17 handicap by 12 strokes, and as such in a match between those two players an allowance should be made for that skill difference to give each player an equal chance of winning. As a matter of reference, the United States Golf Association (USGA) reports that 1.6% of golfers with an official handicap have a handicap rating of 0 or better (reported as a plus handicap, such as +1.2), and these players would be expected to score better than the course rating at least half of the time.

Professionals do not use the handicap system; it is simply that the best score wins. If they were to keep a handicap, it would be significantly to the plus side[^golf-3]. However, players need to be ranked against each other for various purposes.  For example, in 2023, the top 70 players in the FedEx Cup points system on the PGA tour qualify for the PGA Tour playoffs, a system of tournaments to identify a season-long champion for the PGA Tour season.  Also as of this writing the top 30 in the FedEx Cup points system qualify for the PGA Tour's Signature Events, along with those in recent form over the previous several events.

[^golf-3]: In January 2023, Twitter user \@MichaelVerity calculated the current handicap of Jon Rahm to be +13.2.  This means that he is approximately one stroke per hole better than the aforementioned 5 handicapper.

The FedEx Cup points system is how the PGA Tour ranks golfers each year within their organization, but it does not include players who are not members of the PGA Tour.  The PGA Tour, the predominant golf tour in North America, is not the only golf tour in the world. The DP World Tour is primarily in Europe, The Sunshine Tour is primarily in South Africa, the Japanese Tour is the highest level for players there, and there is also an Australasian tour.  The Official World Golf Ranking (OWGR) at owgr.com is a means to try to rank golfers from many different tours in the world.  It normalizes player performances in professional events based on many factor, such as the number of players competing in the tournament and the relative strength of the field based upon the skill levels of the competitors in the event.  When a player is referred to as "the number one player in the world", it generally means that this player is ranked \#1 in the OWGR.  The formula that the OWGR uses has changed over the years since its founding in the 1980s, as have the formulas that the individual tours use (which at first were primarily based upon prize money earned by each player), but attempts to assign points for each professional competing in a sanctioned professional tour no matter where it is located.

While the OWGR started as a project by an agent to rank players for the purpose of sponsorship deals, they are now used in determining the fields for the most prestigious tournaments, in the world. Interestingly, the four most important golf tournaments in the world (the United States Open, the Masters, the PGA Championship, and the Open Championship (also known as the British Open)) are not controlled by any of the professional tours.  Collectively known as the majors, the tournaments are controlled by the USGA, the Augusta National Golf Club, the PGA of America[^golf-4] and the R\&A[^golf-5], respectively.  In determining the fields for these championships, these organizations rely on the OWGR rankings.  For example, at the Masters, the top 50 players in the OWGR at the end of a calendar year are typically issued invitations to the following year's tournament, and the USGA exempts the top 60 in the OWGR from having to qualify for the US Open.  There are many ways to qualify for these tournaments, but these are the primary ways used for professional golfers, particularly at invitationals like The Masters.  Professional golfers on the cusp of the top 50 will try to play well at the end of the year to secure a coveted invite to The Masters.

[^golf-4]: The PGA Tour is a separate entity from The PGA of America, with the Tour having broken off in 1968.

[^golf-5]: The Open Championship was run by the Royal and Ancient Golf Club of Saint Andrews until 2004, when the R&A was formed to support golf around the world and run several championships, including the Open Championship.

In 2022, the Saudi Public Investment Fund founded and funded the LIV golf tour, a worldwide tour with comparatively large prize pools and smaller fields to have golf in both individual and team formats.  Viewing this tour as a competitor, the PGA Tour and DP World Tours gave indefinite suspensions to any player competing in a LIV event.  While most of the 48 players competing on LIV were earning far more money than they would on the PGA or DP World Tours, the one thing they could not earn on the LIV tour were OWGR points.

OWGR points are awarded for playing in tournaments in a rolling two-year basis. A player who won a tournament in October 2020 would continue to keep those points until October 2022.[^golf-6]  The OWGR calculates an average number of points won by adding the number of points earned in the previous two years and dividing it by the number of tournaments played, with a minimum divisor of 40.  While this system works well for players who are continuing to play in events on OWGR-recognized tours, the OWGR did not recognize LIV events from the tour's founding in 2022 through the end of the 2025 season.  Many LIV players qualified for the majors through other means, such as being a past champion with a long-term exemption into that major. For others though, and into other majors, the road was far less clear.  Given that the players had been suspended by the tours where they previously played, which provide the opportunities to earn the most points, the OWGR rankings of the LIV players were bound to drop as the tournaments they played two years prior no longer counted towards their points while their opportunities to earn new points were extremely limited. Moreover, as their collective rankings dropped, the strength of field metrics (as measured by the field's OWGR average rankings) also dropped, creating a vicious circle for the competitors.

[^golf-6]: Note this relationship is constantly changing through this book's writing (2023 to 2026) and may not be current when you read it.  However, the question of this case study remains.

There are many reasons which have been confirmed or suspected as to why the OWGR did not award points to competitors in LIV tournaments for the tour's first four seasons.  Many of these were procedural, such as needing to be in existence for at least a year and completing an application.  Others were perceived as political, such as the composition of the OWGR board of directors being comprised of people like the PGA Tour commissioner who have expressed open disapproval of LIV and their means of signing players to contracts with guaranteed money signing bonuses as well as their disapproval of the LIV's funding source.  Yet another, however, had to do with the nature of the competition itself.  Through 2025, LIV tournaments were conducted over 54 holes of stroke play with no cut and 20\% of the prize pool went to 4-person teams, while most professional tournaments consist of 72 holes of completely individual stroke play where the field is cut approximately in half after 36 holes. Moreover, the tour had no clear way of having new players qualify and underperforming players relegated.

In February 2026, the OWGR finally began awarding points for LIV events, but only to the top 10 finishers (and ties) in each of its 13 individual events, under its "Small Field Tournaments" classification. The OWGR cited LIV's average field size of 57 players, below its minimum of 75. Note that LIV also moved to 72-hole events in 2026. LIV called the top-10 limit "unprecedented," since finishing 11th counts the same as finishing last. For comparison, the winner of LIV's 2026 season opener in Riyadh earned roughly 23 points, less than half of what the winner of a full-field PGA Tour event typically receives.
