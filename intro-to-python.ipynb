### Missing Values {.unnumbered}

We would love to believe that our datasets are always going to be perfectly clean and complete, but in reality, we know that there will always be some data wrangling to do. One common challenge in sports analytics, and analytics in general, is handling missing data. In Python/pandas, missing data is represented by `NaN` (Not a Number) from NumPy, or `None`. When we try to read data and something is missing, pandas marks it as `NaN`. NaNs will show up in your DataFrames as you read and manipulate data. This can be a sign that your calculations were erroneous, and doing more calculations on them may just muddle things more, so be careful! Pandas provides many functions for handling missing data:

- `df.isna()`: Check for missing values
- `df.dropna()`: Remove rows with missing values
- `df.fillna(value)`: Fill missing values with a specified value

## Coding Techniques

### Loops {.unnumbered}

You would be hard-pressed to think of a programming language that does not provide a methodology for repeating a task a specified number of times. One of the more powerful features of computers is their ability to do the same repeated tasks multiple times much more quickly than humans without fatigue or random errors. Python provides `for` loops for repeating tasks. The iterator variable is assigned successive values to execute the loop a stated number of times:

`````python
for i in range(1, 6):
    print(i)
`````

The general form of a `for` loop is:

`````python
#| eval: false

for item in sequence:
    # code to repeat for each item
`````

Sometimes we do not know beforehand the number of iterations we want the loop to run. Suppose we want to add the preface "Andrew competed in the" to our skiing races:

`````python
races = ["downhill", "superg", "downhill", "gs", "slalom",
         "slalom", "dcombined"]

for race in races:
    statement = f"Andrew competed in the {race}"
    print(statement)
`````

Note that Python uses indentation (typically 4 spaces) instead of curly braces to define code blocks. Everything indented under the `for` statement is part of the loop.

### Conditional Evaluation {.unnumbered}

Within Python, sometimes we want our script to do one of two different things depending on the particular situation being evaluated. For example, if the last basket was scored by the home team, then cheer; otherwise, boo. This is what we call a conditional because we are evaluating a situation and executing a block of code or not depending on the results of that condition. The type of condition we just described above is called an if-then-else structure:

`````python
#| eval: false

if logical_condition:
    # execute code if the condition is true
# Continue with the rest of the script
`````

Python evaluates the condition. If true, the indented code block executes. If false, the program continues after the if block. We can add alternative instructions with the `else` clause:

`````python
#| eval: false

if condition:
    # execute code that runs when the condition is true
else:
    # execute code that runs when the condition is false
# Continue with the rest of the script
`````

You can nest `if` statements using `elif` (else if):

`````python
#| eval: false

if condition_1:
    # execute code that runs when condition1 is true
elif condition_2:
    # execute code when condition1 is false but condition2 is true
else:
    # execute code when both condition1 and condition2 are false
# Continue with the rest of the script
`````

In addition to if-then-else constructs, Python supports the evaluation of code while a condition is met using the `while` loop. Be careful when using while logic, as you can easily get stuck in a never-ending loop!

`````python
#| eval: false

while logical_condition:
    # Only execute this code while the logical_condition is TRUE
`````

`````python
# Load team batting data for multiple years (2010-2025)
from mlb_utils import team_batting_mlb

years_data = []
for year in range(2010, 2026):
    years_data.append(team_batting_mlb(year))

# Combine all years into one DataFrame
multi_year_batting = pd.concat(years_data, ignore_index=True)

# Filter for specific teams
df_filtered = multi_year_batting[multi_year_batting['Team'].isin(['COL', 'MIA'])].copy()

# Select and rename columns for clarity
team_hr = df_filtered[['Season', 'Team', 'HR']].copy()
team_hr = team_hr.rename(columns={'Team': 'teamID', 'Season': 'yearID', 'HR': 'team_hr'})

print(team_hr.head())
print(f"\nTotal years of data: {team_hr['yearID'].nunique()}")
print(f"\nMost recent season data:")
print(team_hr[team_hr['yearID'] == 2025])
`````

Pandas supports method chaining, where we apply multiple operations in sequence.

### Debugging Your Code {.unnumbered}

![](fig/bugs.jpeg){width=35%}

Like in all programming languages, eventually, you are going to make mistakes. These mistakes could be syntax errors, where the interpreter cannot understand your code, or logic errors, where your code runs but gives you output that is incorrect for what you are hoping to analyze.

Unfortunately, this is one of the most frustrating parts of coding, as finding the source of these errors and fixing them is not always obvious. Python provides helpful error messages called "tracebacks" that show the sequence of function calls that led to an error. Our recommendation for handling errors that do not have an obvious fix is to begin by searching the Internet for that error message. There is a reasonable chance that you will be able to find how others fixed this error, particularly on sites like StackOverflow.com. In your search term, include as much detail as you can from the error message, but to get the best match, you might refrain from including variable names specific to your code.

If you can identify where the error is in your code from the traceback, go to that section and start there. Python might not tell you exactly where your error is occurring, because it will try to run through as many lines of code as it can before finally refusing to run.

Once you've found the error but see that the fix is not obvious, you may try several strategies to isolate the error. This is often a process of trial and error and may be quite frustrating, but debugging is partly detective work, and you may have to begin eliminating possibilities. One way to do this is to try to rewrite the section of code where you are getting errors and then introduce bits of complexity line by line. Instead of deleting troublesome sections of code, try commenting out lines by using the hashtag (`#`) and then run your code to see if you are getting the expected results.

As you add lines of code, you might use `print()` to print values to the console, and that way, you can check to see if they match what you expect. You can use `df.head()` to see if the DataFrame you are expecting is really what the Python interpreter is examining.

Moreover, on the occasions where your instructor allows this, don't be afraid to ask one of your classmates! It's possible that they have encountered similar errors and know how to fix them.

3. Plot the total number of wins by year for each team. Use data from 1990 to present.

`````python
#| eval: false
#| include: false

import matplotlib.pyplot as plt
import seaborn as sns

# Get multiple years of data
years_data = []
for year in range(1990, 2026):
    pitch = pitching_stats_mlb(year)
    pitch['Season'] = year
    years_data.append(pitch)

multi_year = pd.concat(years_data, ignore_index=True)
teams_multi = multi_year[multi_year['Team'].isin(['COL', 'NYY'])].copy()

# Aggregate wins by team and year
wins_by_year = (teams_multi
    .groupby(['Team', 'Season'])
    .agg({'W': 'sum'})
    .reset_index()
)

# Plot
sns.lineplot(data=wins_by_year, x='Season', y='W',
             hue='Team', marker='o')
plt.xlabel('Year')
plt.ylabel('Total Wins')
plt.title('Pitching Wins by Year')
plt.grid(True, alpha=0.3)
plt.show()
`````

Let us now shift gears to batting data. Slugging percentage (SLG) is a baseball metric that computes a player's total bases per at bat. Total bases is defined as

$$
TB = singles + 2 \times doubles + 3 \times triples + 4 \times home\ runs
$$

4. Create new columns for Total Bases (TB) and Slugging Percentage (SLG) for the two teams chosen above.

`````python
#| eval: false
#| include: false

from mlb_utils import batting_stats_mlb

df_bat = batting_stats_mlb(2025)
df_bat = df_bat[df_bat['Team'].isin(['COL', 'NYY'])].copy()

# Calculate singles, TB, and SLG
df_bat['1B'] = df_bat['H'] - df_bat['2B'] - df_bat['3B'] - df_bat['HR']
df_bat['TB'] = df_bat['1B'] + 2*df_bat['2B'] + 3*df_bat['3B'] + 4*df_bat['HR']
df_bat['SLG'] = df_bat['TB'] / df_bat['AB']

# Remove any rows with NaN
df_bat = df_bat.dropna(subset=['SLG'])
`````

5. Plot the average `SLG` for one of your teams from 2015 through 2025.

`````python
#| eval: false
#| include: false

# Filter for one team
col_df = df_bat[df_bat['Team'] == 'COL'].copy()

# Calculate average SLG (per player)
avg_slg = (col_df
    .groupby('Season')
    .agg({'SLG': 'mean'})
    .reset_index()
)

# Plot
plt.plot(avg_slg['Season'], avg_slg['SLG'], marker='o', color='black')
plt.xlabel('Year')
plt.ylabel('Average Slugging Percentage')
plt.title('Average SLG by Year: Colorado Rockies')
plt.grid(True, alpha=0.3)
plt.show()
`````

6. Note that question 5 is asking for the average `SLG` and not the team `SLG`. So now, plot the team `SLG` by year since 2015.

`````python
#| eval: false
#| include: false

# Calculate team SLG (total TB / total AB)
team_slg = (col_df
    .groupby('Season')
    .agg({'TB': 'sum', 'AB': 'sum'})
    .reset_index()
)
team_slg['team_SLG'] = team_slg['TB'] / team_slg['AB']

# Plot
plt.plot(team_slg['Season'], team_slg['team_SLG'], marker='o', color='black')
plt.xlabel('Year')
plt.ylabel('Team Slugging Percentage')
plt.title('Team SLG by Year: Colorado Rockies')
plt.grid(True, alpha=0.3)
plt.show()
`````
