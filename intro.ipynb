## Why Sports Analytics?

If you are reading this book, you almost certainly have an interest in sports. People have engaged in sport for thousands of years, and the ancient Olympic Games that inspired our modern Olympics date back almost 3,000 years. As long as there have been athletes and competitions, there have always been those seeking to gain an edge. Some are accepted within the rules and spirit of sport, like better training methods and calendars and nutrition and technology, while others violate the rules, like secretly using equipment that has been deemed to be nonconforming with the rules of the sport or having competitors (both human and animal) use performance-enhancing drugs.

As data and computing became more plentiful and available, people turned to analytics to gain an advantage in sport. It is unclear who was the first sports analyst or who was the first team or sport to employ an analyst like we might think of today, but it is quite clear which two major inflection points caused analytics to gain traction. One of these inflection points was developed by an insider; the other was a revelation by an outsider.

1. 1977: Bill James, a graduate of the University of Kansas and a freelance writer, publishes the 1977 Baseball Abstract: Featuring 18 Categories of Statistical Information That You Just Can't Find Anywhere Else. James used his unique (at the time) statistical analysis from box scores of the previous year's baseball games. While this self-published book sold fewer than 100 copies, it set in motion a new means for fans to analyze the performance of players and teams called sabermetrics. While many fans loved this, baseball scouts and managers, at least publicly, did not take to this new means of analysis. However, a quarter century later, it caused at least one team to think about this differently.

![The cover of the 1988 edition of The Bill James Baseball Abstract.](fig/james-abstract.jpeg){#fig-baseball-almanac}


2. 2003: Michael Lewis's book Moneyball is published. Lewis's book detailed the adoption of baseball analytics techniques by the 2002 Oakland A's baseball team as they tried to rebuild from losing three players from their stellar 2001 season to teams with more resources who could pay the players more. He detailed general manager Billy Beane's pursuit of the art of winning an unfair game by using advanced analytics to identify value in players that other teams did not see. His techniques were first ridiculed by other teams and then slowly adopted. By the time the 2011 movie of the same name was released, starring Brad Pitt as Billy Beane, every team in Major League Baseball had some sort of an on-the-field analytics department.

![The cover of Michael Lewis' book, Moneyball.](fig/moneyball.jpeg){#fig-moneyball width=35%}

If you are reading this textbook, you are probably familiar with Billy Beane, Brad Pitt, and the movie *Moneyball* even if you have never heard of Bill James or Michael Lewis. Maybe that is what inspired you to look at sports through an analytical lens. Maybe you were just tired of finishing in last place in your annual fantasy football league. Maybe you need a good excuse to learn Python programming and improve your mathematical modeling skills. Or maybe you have both a love of sport and an aptitude for data and mathematics and want to conspire to put the two together simply to be a more informed fan. Regardless of your personal reasons, welcome to the world of sports analytics.

## What Is Sports Analytics?

Sports analytics is the use of data and advanced mathematical methods to build models to help a player or team gain a competitive advantage on the field of play. The field of play designation is important in this context, because this subject does not encompass the analytics that a team might use to determine ticket packages, hot dog pricing, or other types of interactions with their customer base that any other entertainment product like a ballet company or theater might use.

While sports and sports reporters have long used data and frequency counts to report on players and games and the state of the team/sport, in general, sports analytics will go well beyond counting the number of home runs or calculating a free throw percentage or a goalie's save percentage. Sports analysts certainly use these statistics, and they often develop more statistics like the wins above replacement (WAR) player rating in baseball or the Corsi statistic in hockey. The difference between statistics and analytics is that analytics involves building models that are composed of statistics, and the models can be used to inform a strategy or select players in search of a competitive advantage over opponents.

## Careers in Sports Analytics

Sports analytics is a highly sought-after career field. Like many jobs in sports, including players, there is a far higher demand for jobs with teams than there is a supply of the jobs. The pay structure is relatively top-heavy, also similar to the players, where a small number of stars make the majority of the salary pool and a large number of supporting cast members are working for far less than might be earned using similar analytic skills in a different field of business. Yet there is still the allure of working directly for a team that, for many people, far outweighs the compensation in salary. Fortunately, to help with the demand, there are many other jobs in sports analytics that use the tool sets we teach in this textbook but do not require that you work directly for a team.

Working for a team is the dream job for many in sports analytics. The team analyst uses data to evaluate player selection, game strategy, and situational decision-making. Jonah Hill's character Peter Brand in Moneyball might be the quintessential team analyst. While his position was unique at the time, most teams have some sort of analytics department, often with multiple analysts.

## Off-the-Field Analytics

This book focuses on the analytics decisions that are made by individuals and teams directly regarding on-the-field play. Many of these same techniques can be used by sports teams and organizations for off-the-field decisions. For example, how much should a team charge for a Friday night ticket when they are playing the New York Yankees as opposed to a Wednesday afternoon game against the Pittsburgh Pirates? How should a team market itself to get more fans to come to the games, where they are more likely to buy $15 beers and $10 hot dogs and pay $30 for parking, often rendering the ticket price itself inconsequential? Should a team be willing to play a college football game on a Tuesday night in order to get national TV exposure when they might not normally get it in the usual Saturday afternoon spot?

All of these are very important to the financial success of a team, and in general, the larger the attendance at a game, the greater the home-field advantage, which translates somewhat to on-the-field success. However, these analytics are beyond the scope of this textbook.

### Research Block: Do Players Get Streaky? {.unnumbered}

In each chapter, we will include a discussion about a piece of research referencing sports analytics. These will be summaries of academic studies that examine a sports-related question or a theoretical question that is addressed through the lens of sports. In this chapter, we introduce one for thought about randomness and streaks and cognitive bias.

Thomas Gilovich is a psychology professor at Cornell University specializing in research on heuristics and cognitive biases. As part of his early research, he was concerned with individuals' abilities to understand randomness in sequences. An avid sports fan whose doctoral dissertation studied the concepts of biased evaluation and persistence within the context of gambling, it was a natural fit to use sports as the domain for studying cognitive biases.

In his 1985 paper with Robert Vallone and Amos Tversky, "The Hot Hand in Basketball: On the Misperception of Random Sequences," which can be found at [https://www.sciencedirect.com/science/article/abs/pii/0010028585900106](https://www.sciencedirect.com/science/article/abs/pii/0010028585900106), he conducted a series of studies involving both individuals' perceptions in a survey and statistical analysis while looking for evidence of a "hot hand" or even a "cold hand." The hot hand is used to represent the idea that a player who makes a basket is more likely to make the next basket than his base rate (think: skill level) would imply, while the cold hand implies one who misses will continue to miss at a greater-than-base rate. He examined data from both the Cornell University men's basketball team and the Philadelphia 76ers NBA team. Despite the 76ers roster including Andrew Toney, a player who was known in the media as a notorious streak shooter, his analysis found no evidence of either a hot or a cold hand[^1].

![Andrew Toney on the Philadelphia 76ers in the early 1980s.](fig/toney.jpeg){#fig-toney width=35%}

Gilovich concluded through his study that the hot hand effect is simply an illusion that is a limitation of people's memories. Individuals are more likely to remember a streak but yet are unable to comprehend the frequency with which streaks appear in random sequences. This, however, does not stop people from believing in a hot hand, even in contexts as random as roulette. Casinos take advantage of this by posting the previous results of the spins on a signboard next to the roulette wheel, encouraging people to place their bets and implying that people might want to bet against a streak continuing to occur, as there is no way that a ball could land in a red pocket yet again if it had landed in a red one on the previous six spins. (If you are unfamiliar with roulette, essentially, it is a coin flip if the ball lands in a red or a black pocket.) The casinos use a similar strategy in another essentially coin-flip game, baccarat.

![A signboard at a roulette table.](fig/roulette.jpeg){#fig-roulette width=35%}

The hot hand theory continues to be tested in a variety of domains and sports, including baseball, soccer, volleyball, bowling, stock market investing, and even dice games and lotteries. Indeed, the authors of this textbook published a paper in 2018 examining this effect in golf. In the nearly 40 years since Gilovich published this initial paper, the power of sports analytics not only to answer sports questions but to inform more general topics continues to expand with the volume of data available.

[^1]: Interestingly, the only player whose current shot success rate was statistically dependent on the previous shot was Darryl Dawkins, and it was in the opposite direction (i.e., he was more likely than the base rate would imply to miss the current shot if he had made the previous one, and vice versa).

### Case Study: Transitioning to Analytic Thinking in Sports {.unnumbered}

Each chapter will also include a case study relative to the topic of the chapter. Sometimes there will be direct coding exercises related to the case study, and other times there will be topics for discussion. Major organized professional team sports have existed for over 100 years. During that time, there have been many methods that teams have used to evaluate potential players for those teams. Some sports have the benefit of colleges or schools to do the training, while others (particularly soccer) have created their own junior divisions of their clubs to handle training internally---sometimes of children of single-digit ages. In this case study, we will focus largely on evaluating players who have reached adulthood but have not yet been made a part of a team in a top-level league.

In sports like baseball, for over a century, major-league teams have relied on scouts to go to parts of the country and view players in high school, college, American Legion, Pony, or other organized leagues and summer leagues to identify if there are players who should be considered for signing as a part of their clubs. Increasingly, MLB teams look internationally for players, with the Dominican Republic being a hot spot for academies for teams to look for talent. These scouts will write reports based on their observations of a player and send them to team management.

![Reproduction of Ken Griffey Jr.'s scouting report while attending Moeller High School in Cincinnati, Ohio (1987).](fig/kg-scouting-report.jpg){#fig-griffey width=50%}

In American football, in addition to scouting reports, the NFL also holds the NFL Scouting Combine, a set of workouts in the late winter in Indianapolis for college players who are eligible for the draft. At this combine, players are examined in a variety of running and jumping drills, and many parts of their bodies are officially measured. There are other examples for other sports, but it is under these watchful eyes that many of the evaluations are made.

The years of experience of the scouts are often long, and their expertise is valuable. However, scouts can be prone to human bias, falling in love with someone who "looks the part" and disregarding talent who doesn't have that look. At the end of the 20th century, one needs to look no further than quarterback Ryan Leaf, who was the 2nd player chosen in the 1998 NFL draft. At 6' 5.5" (196 cm) and 235 pounds (106 kg), and recently completing a brilliant junior season at Washington State University, he had the look of the prototypical quarterback. The Indianapolis Colts had the first pick of the draft that year and ultimately chose Peyton Manning, who was a similar size and had an extremely successful college career. Manning ultimately won five NFL MVP awards and two Super Bowls and was inducted into the NFL Hall of Fame in 2021, while Leaf played in a total of 25 games over four years and is widely regarded as one of the biggest NFL "busts" of all time. Moreover, two years later, it was not until the 199th pick that the New England Patriots selected Tom Brady in the draft. While tall, he was considered slight of build (for an NFL quarterback) and slow and had a mediocre collegiate career. However, he played 22 NFL seasons; won a total of seven Super Bowls, five Super Bowl MVPs, and three NFL MVP awards; and is often referred to as "The GOAT"---an acronym meaning "Greatest of All Time."

Scouting is not a perfect science, of course, and there can always be some freak injury that derails a highly regarded prospect or someone with little exposure from a small school who winds up being quite successful. Analytics offer a tool set that can be used for player evaluation that can help teams not be blinded by a human bias. While the models that teams use for player evaluation may include some of the data from scouting reports, they may also include other inputs, many of which could be proprietary to the team. Analytic models are also designed to reduce the influence of human bias. This does not mean that a team should always defer to whatever answer the computer spits out, but a model's prediction of player success is another data point that a team can use, along with contract considerations, for example.
